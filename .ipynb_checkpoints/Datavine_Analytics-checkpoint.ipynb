## Executive Summary
This notebook presents three prototype machine learning solutions developed for our boutique consulting clients. The project encompasses the complete data science workflow—from exploratory data analysis and preprocessing to model implementation and evaluation. 

The three strategic initiatives are:
1. **Wine Classification System:** A k-Nearest Neighbors (k-NN) model optimized with PCA for automated wine classification.
2. **Agricultural Feed Recommendation Engine:** A similarity-based recommendation system using PCA to suggest comparable poultry feeds.
3. **Regional Crime Pattern Analysis:** A clustering approach using K-Means and Gaussian Mixture Models (GMM) to discover natural groupings in public safety data.

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture

# Set visualization style
sns.set_theme(style="whitegrid")
np.random.seed(42)
```

## Step 1: Data Preparation and Overview
Robust machine learning models require clean, well-structured data. In this step, we load the Wine, Chickwts, and USArrests datasets. We inspect them for missing values, standardize the numerical features to ensure uniform scaling, and summarize their structures to confirm readiness for downstream modeling.

```python
# 1. Load Datasets
# Wine Dataset (Classification)
wine_raw = load_wine(as_frame=True)
wine_df = wine_raw.frame
wine_df.rename(columns={'target': 'wine_class'}, inplace=True)

# Chickwts Dataset (Recommendation)
chickwts_df = pd.read_csv('chickwts.csv').drop(columns=['Unnamed: 0'], errors='ignore')

# USArrests Dataset (Clustering)
usarrests_df = pd.read_csv('USArrests.csv').rename(columns={'Unnamed: 0': 'State', 'rownames': 'State'})
if 'State' in usarrests_df.columns:
    usarrests_df.set_index('State', inplace=True)

# 2. Inspect Missing Values
print("Missing Values Check")
print(f"Wine missing values: {wine_df.isnull().sum().sum()}")
print(f"Chickwts missing values: {chickwts_df.isnull().sum().sum()}")
print(f"USArrests missing values: {usarrests_df.isnull().sum().sum()}")

# 3. Dataset Structural Summary
print("\nDataset Summaries")
print(f"Wine Dataset: {wine_df.shape[0]} samples, {wine_df.shape[1]-1} features, {wine_df['wine_class'].nunique()} classes")
print(f"Chickwts Dataset: {chickwts_df.shape[0]} samples, {chickwts_df['feed'].nunique()} distinct feed types")
print(f"USArrests Dataset: {usarrests_df.shape[0]} regions, {usarrests_df.shape[1]} features")
```

## Step 2: Wine Classification System
**Client:** Premium Wine Distributor
**Objective:** Automatically classify wines based on chemical properties for inventory management.

We apply Principal Component Analysis (PCA) to reduce dimensionality while retaining 95% of the variance. We then use `GridSearchCV` to systematically tune the hyperparameters of a k-Nearest Neighbors (k-NN) classifier to identify the optimal configuration.

```python
# 1. Feature Separation
X_wine = wine_df.drop(columns=['wine_class'])
y_wine = wine_df['wine_class']

# Split dataset
X_train_w, X_test_w, y_train_w, y_test_w = train_test_split(
    X_wine, y_wine, test_size=0.3, random_state=42, stratify=y_wine
)

# 2. Standardization
scaler_wine = StandardScaler()
X_train_w_scaled = scaler_wine.fit_transform(X_train_w)
X_test_w_scaled = scaler_wine.transform(X_test_w)

# 3. PCA (Retaining 95% variance)
pca_wine = PCA(n_components=0.95, random_state=42)
X_train_w_pca = pca_wine.fit_transform(X_train_w_scaled)
X_test_w_pca = pca_wine.transform(X_test_w_scaled)

print(f"Original feature count: {X_train_w_scaled.shape[1]}")
print(f"PCA components retaining 95% variance: {pca_wine.n_components_}")

# 4. Hyperparameter Tuning using GridSearchCV
param_grid = {
    'n_neighbors': list(range(1, 15)),
    'metric': ['euclidean', 'manhattan', 'minkowski'],
    'weights': ['uniform', 'distance']
}

grid_search = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train_w_pca, y_train_w)
best_knn = grid_search.best_estimator_

print(f"\nOptimal Hyperparameters: {grid_search.best_params_}")

# 5. Model Evaluation
y_pred_w = best_knn.predict(X_test_w_pca)
print(f"Test Accuracy: {accuracy_score(y_test_w, y_pred_w):.4f}\n")
print("Classification Report:")
print(classification_report(y_test_w, y_pred_w, target_names=wine_raw.target_names))
```

## Step 3: Agricultural Feed Recommendation Engine
**Client:** Agricultural Supply Company
**Objective:** Suggest comparable feed products based on historical performance.

To build the recommendation engine, we aggregate weight statistics per feed type, standardize the measurements, and extract a primary principal component. We then compute the cosine similarity to rank and recommend alternative feeds that yield similar growth trajectories.

```python
# 1. Standardize Weight Feature
scaler_chick = StandardScaler()
chickwts_df['weight_scaled'] = scaler_chick.fit_transform(chickwts_df[['weight']])

# 2. Build Feed Profiles (Aggregated Statistics)
feed_profiles = chickwts_df.groupby('feed')['weight_scaled'].agg(
    ['mean', 'std', 'median', 'min', 'max']
).fillna(0)

# 3. PCA to 1 Principal Component
pca_feed = PCA(n_components=1, random_state=42)
feed_pca = pca_feed.fit_transform(feed_profiles)
feed_representation = pd.DataFrame(feed_pca, index=feed_profiles.index, columns=['PC1'])

# 4. Compute Cosine Similarity
cos_sim_matrix = cosine_similarity(feed_representation)
cos_sim_df = pd.DataFrame(cos_sim_matrix, index=feed_profiles.index, columns=feed_profiles.index)

# 5. Recommendation Generator
def recommend_feeds(feed_name, top_n=2):
    return cos_sim_df[feed_name].drop(feed_name).sort_values(ascending=False).head(top_n)

# Display Recommendations
print("Recommended Alternative Feeds")
for feed in cos_sim_df.index:
    recs = recommend_feeds(feed)
    formatted_recs = ", ".join([f"{idx} (Sim: {val:.2f})" for idx, val in recs.items()])
    print(f"Alternatives for {feed.capitalize().ljust(10)} -> {formatted_recs}")
```

## Step 4: Regional Crime Pattern Analysis
**Client:** Public Policy Research Firm
**Objective:** Identify natural groupings in regional crime statistics.

We focus on the top three violence-related features (Murder, Assault, Rape), standardizing and reducing them to 2 Principal Components for visualization. We utilize both the Elbow Method (for K-Means) and Bayesian Information Criterion (for GMM) to determine the optimal number of clusters before modeling.

```python
# 1. Feature Selection & Standardization
selected_features = ['Murder', 'Assault', 'Rape']
X_crime = usarrests_df[selected_features]

scaler_crime = StandardScaler()
X_crime_scaled = scaler_crime.fit_transform(X_crime)

# 2. PCA to 2 Components
pca_crime = PCA(n_components=2, random_state=42)
X_crime_pca = pca_crime.fit_transform(X_crime_scaled)
pca_df = pd.DataFrame(X_crime_pca, columns=['PC1', 'PC2'], index=usarrests_df.index)

# 3. Optimal Cluster Selection
k_range = range(1, 9)
inertia_scores, bic_scores = [], []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X_crime_pca)
    inertia_scores.append(km.inertia_)
    
    gmm = GaussianMixture(n_components=k, random_state=42).fit(X_crime_pca)
    bic_scores.append(gmm.bic(X_crime_pca))

# 4. Apply Final Models (Selected K = 3 based on standard elbow/BIC)
optimal_k = 3

kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
pca_df['KMeans_Cluster'] = kmeans_final.fit_predict(X_crime_pca)

gmm_final = GaussianMixture(n_components=optimal_k, random_state=42)
pca_df['GMM_Cluster'] = gmm_final.fit_predict(X_crime_pca)

# 5. Visual Comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.scatterplot(data=pca_df, x='PC1', y='PC2', hue='KMeans_Cluster', palette='Set1', s=80, ax=axes[0])
axes[0].set_title('K-Means Hard Clustering')

sns.scatterplot(data=pca_df, x='PC1', y='PC2', hue='GMM_Cluster', palette='Set2', s=80, ax=axes[1])
axes[1].set_title('GMM Probabilistic Clustering')

plt.tight_layout()
plt.show()
```
